## Objectives

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats

%matplotlib inline
plt.style.use('ggplot')
```

## 1. Data Preparation

Load and inspect the match data.

```python
# Example: generate synthetic match data
np.random.seed(42)
n_matches = 38

data = pd.DataFrame({
    'match_week': range(1, n_matches + 1),
    'home_goals': np.random.poisson(lam=1.5, size=n_matches),
    'away_goals': np.random.poisson(lam=1.1, size=n_matches),
    'home_xG': np.round(np.random.normal(1.5, 0.6, n_matches), 2),
    'away_xG': np.round(np.random.normal(1.1, 0.5, n_matches), 2),
})

data['goal_diff'] = data['home_goals'] - data['away_goals']
data.head(10)
```

## 2. Descriptive Statistics

```python
print('Summary Statistics')
print('=' * 40)
print(data[['home_goals', 'away_goals', 'home_xG', 'away_xG']].describe())

print(f"\nHome win rate: {(data['goal_diff'] > 0).mean():.1%}")
print(f"Draw rate:     {(data['goal_diff'] == 0).mean():.1%}")
print(f"Away win rate: {(data['goal_diff'] < 0).mean():.1%}")
```

## 3. Visualisation

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

# Goal distribution
axes[0].hist(data['home_goals'], bins=range(0, 8), alpha=0.6, label='Home', color='#3498db', edgecolor='black')
axes[0].hist(data['away_goals'], bins=range(0, 8), alpha=0.6, label='Away', color='#e74c3c', edgecolor='black')
axes[0].set_xlabel('Goals')
axes[0].set_ylabel('Frequency')
axes[0].set_title('Goal Distribution')
axes[0].legend()

# xG comparison scatter
axes[1].scatter(data['home_xG'], data['away_xG'], alpha=0.7, edgecolors='black')
lim = max(data['home_xG'].max(), data['away_xG'].max()) + 0.5
axes[1].plot([0, lim], [0, lim], 'k--', alpha=0.4)
axes[1].set_xlabel('Home xG')
axes[1].set_ylabel('Away xG')
axes[1].set_title('Home vs Away xG')

# Goal difference over time
axes[2].bar(data['match_week'], data['goal_diff'],
            color=data['goal_diff'].apply(lambda x: '#2ecc71' if x > 0 else ('#e74c3c' if x < 0 else '#95a5a6')))
axes[2].axhline(0, color='black', linewidth=0.8)
axes[2].set_xlabel('Match Week')
axes[2].set_ylabel('Goal Difference')
axes[2].set_title('Goal Difference by Match Week')

plt.tight_layout()
plt.show()
```

## 4. Statistical Tests

### 4.1 Sign Test – Home Advantage

```python
from plot_Sign_Test import sign_test, plot_sign_test_results

sign_result = sign_test(data['home_goals'].values, data['away_goals'].values)
print('Sign Test Results:')
for k, v in sign_result.items():
    print(f'  {k}: {v}')

plot_sign_test_results(sign_result, title='Sign Test: Home vs Away Goals')
```

### 4.2 T-Test – Expected Goals Comparison

```python
from plot_t_test import perform_t_test, plot_t_test_results

t_result = perform_t_test(data['home_xG'].values, data['away_xG'].values, test_type='paired')
print('T-Test Results:')
for k, v in t_result.items():
    print(f'  {k}: {v}')

plot_t_test_results(
    data['home_xG'].values, data['away_xG'].values,
    t_result,
    labels=('Home xG', 'Away xG'),
    title='Paired T-Test: Home vs Away xG'
)
```
